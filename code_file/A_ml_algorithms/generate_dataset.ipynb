# Combination with individual simple feature

```python
df_densenet = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\code_file\densenet_feature_extraction\output\densenet121_features_notest.csv')
df_basic_ad = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\basic_ad_features_all_disease.csv')
df_basic = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\basic_features_all_diseases.csv')
df_gist = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\gist_features_all_diseases.csv')
df_glcm = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\glcm_features_all_diseases.csv')
df_hist = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\hist_features_all_disease.csv')
df_lbp = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\lbp_features_all_diseases.csv')
df_densenet.shape, df_basic_ad.shape, df_basic.shape, df_gist.shape, df_glcm.shape, df_hist.shape, df_lbp.shape
```

```python
# Danh sách các handcrafted features để lặp qua
handcrafted_dict = {
    'basic_ad': df_basic_ad,
    'basic': df_basic,
    'gist': df_gist,
    'glcm': df_glcm,
    'hist': df_hist,
    'lbp': df_lbp
}

# Dictionary để lưu trữ các DataFrame sau khi ghép
fusion_dfs = {}

print("--- Quá trình thực thi Concatenate ---")

for name, df_hc in handcrafted_dict.items():
    # Bước 2 & 3: Loại bỏ cột đầu tiên của DataFrame handcrafted (iloc[:, 1:])
    # Việc này không ảnh hưởng đến biến gốc nếu bạn cần dùng lại
    df_hc_cleaned = df_hc.iloc[:, 1:]
    
    # Bước 1: Ghép nối với DenseNet theo axis=1
    # Kết quả sẽ bao gồm toàn bộ cột của DenseNet + các cột đặc trưng (trừ cột đầu) của df handcrafted
    df_combined = pd.concat([df_densenet, df_hc_cleaned], axis=1)
    
    # Lưu vào dictionary với tên tương ứng
    fusion_dfs[f'df_fusion_{name}'] = df_combined
    
    print(f"Đã tạo: df_fusion_{name:8} | Shape mới: {df_combined.shape}")

# Sau khi chạy xong, bạn có thể truy cập từng DataFrame như sau:
# df_fusion_gist = fusion_dfs['df_fusion_gist']
# df_fusion_lbp = fusion_dfs['df_fusion_lbp']

print("\nHoàn tất! Bạn có thể sử dụng các DataFrame trong 'fusion_dfs' để huấn luyện mô hình.")
```

```python
import os

# Đường dẫn thư mục bạn muốn lưu các file fusion
# Tôi sẽ tạo một thư mục 'fusion_results' trong project của bạn để dễ quản lý
output_dir = r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\code_file\A_feature_selection\csv_data\fusion_results'

# Tạo thư mục nếu chưa tồn tại
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"Đã tạo thư mục: {output_dir}")

print("--- Bắt đầu lưu file CSV ---")

for name, df_fusion in fusion_dfs.items():
    # Tạo tên file: ví dụ 'df_fusion_gist.csv'
    file_path = os.path.join(output_dir, f"{name}.csv")
    
    # Lưu file (index=False để không lưu thêm cột chỉ số của pandas)
    df_fusion.to_csv(file_path, index=False)
    
    print(f"Đã lưu thành công: {file_path}")

print("\n--- Hoàn tất! Tất cả các bộ đặc trưng đã sẵn sàng ---")
```

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\healthy_leaf\hist_features_healthy_leaf.csv')
df2 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\insect_Pest_Disease\hist_features_insect_pest_disease.csv')
df3 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\Leaf_Spot_Disease\hist_features_leaf_spot_disease.csv')
df4 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\Mosaic_Virus_Disease\hist_features_mosaic_virus_disease.csv')
df5 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\White_Mold_Disease\hist_features_white_mold_disease.csv')
df6 = pd.read_csv(r'E:\PROJECTWORSHOP\Eggplant Leaf Disease Detection Dataset\csv_data\Wilt_Disease\hist_features_wilt_disease.csv')
df1.shape, df2.shape, df3.shape, df4.shape, df5.shape, df6.shape
```

```python
df_concat = pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)
df_concat.head(), df_concat.shape
df_concat.to_csv('hist_features_disease.csv', index=False)
```

```python
label_col = 'Name' if 'Name' in df_concat.columns else df_concat.columns[0]

label_counts = df_concat[label_col].value_counts().reset_index()
label_counts.columns = ['label', 'Number of Samples']

label_counts
```
